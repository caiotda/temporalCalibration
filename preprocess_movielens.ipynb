### Prepare data for cross validation


In this section, we divide data as follows:

70%: train set
15%: test set
15%: val set


Hyperparameter tuning will be evaluated on the validation set.

```python
X = movielens.drop(columns=['timestamp', 'genres_list', 'rating'])
```

```python
y = movielens[["rating"]]
```

```python
X_train, X_test_val, y_train, y_test_val = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_test_val
```

```python
X_test, X_val, y_test, y_val = train_test_split(X_test_val, y_test_val, test_size=0.5, random_state=42)
```

### Saving features

```python
save(X_train, TRAIN_FEATURES_PATH)
save(X_test, TEST_FEATURES_PATH)
save(X_val, VAL_FEATURES_PATH)
```

### Saving targets

```python
save(y_train, TRAIN_TARGET_PATH)

save(y_test, TEST_TARGET_PATH)

save(y_val, VAL_TARGET_PATH)
```
